# 阶段 2 · 步骤 1：用 NetworkX 建第一个图

学习目标：
1. 图 = 节点 + 边，边可以带权重
2. `G.edges()` 遍历边、`G.neighbors(n)` 查邻居、`G[u][v]` 访问边属性
3. `spring_layout` 给节点算坐标，`nx.draw` 负责画

逐 cell 往下走，每个 cell 运行后观察输出。

```python
import random

import matplotlib.pyplot as plt
import networkx as nx

print('imports OK')
```

## 1. 建图

`nx.Graph()` 是无向图。`add_edges_from` 一次加多条边，节点自动创建（编号 0-4）。

```python
G = nx.Graph()
G.add_edges_from([(0, 1), (1, 2), (2, 3), (3, 4), (0, 4), (1, 3)])
print(f"节点数: {G.number_of_nodes()}")
print(f"边数: {G.number_of_edges()}")
```

## 2. 给边加随机权重

`random.uniform(a, b)` 返回 `[a, b]` 区间的小数——权重应该是小数，不是整数（新手常见错误：用 `randint`）。

`G[u][v]` 直接访问边属性字典，赋值 `'weight'` 即可。

```python
for u, v in G.edges():
    G[u][v]['weight'] = round(random.uniform(0.1, 1.0), 2)

# 最常用的边遍历写法：G.edges(data=True) 同时拿到两端点和属性字典
for u, v, d in G.edges(data=True):
    print(f"边 ({u}, {v}) 权重 = {d['weight']}")
```

## 3. 查邻居

注意：`G.neighbors(0)` 返回**迭代器**，直接 print 只显示对象地址，必须 `list()` 转换才能看内容。

这个坑以后会反复遇到（`G.edges` 也是迭代器）。

```python
print(f"节点 0 的邻居: {list(G.neighbors(0))}")
```

## 4. 可视化

- `spring_layout`：力导向布局，算出每个节点的坐标；`seed=42` 固定布局，每次运行图形一样
- `nx.draw`：画节点和边
- `draw_networkx_edge_labels`：把权重标在边上

```python
pos = nx.spring_layout(G, seed=42)
nx.draw(G, pos, with_labels=True, node_color='skyblue', node_size=700, font_weight='bold')
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)
plt.show()
```
